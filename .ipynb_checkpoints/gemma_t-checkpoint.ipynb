```python
# Common imports
import os
import jax
import jax.numpy as jnp

# Gemma imports
from gemma import gm

# Allocate full GPU memory
os.environ["XLA_PYTHON_CLIENT_MEM_FRACTION"] = "1.00"

# Load Gemma 3 4B model
model = gm.nn.Gemma3_4B()

# Load checkpoint parameters
params = gm.ckpts.load_params(
    gm.ckpts.CheckpointPath.GEMMA3_4B_IT
)

# Create chat sampler
sampler = gm.text.ChatSampler(
    model=model,
    params=params,
    multi_turn=True,
    print_stream=True,
)

# Generate response
turn0 = sampler.chat(
    'Share one metaphor linking "shadow" and "laughter".'
)

print(turn0)
```
